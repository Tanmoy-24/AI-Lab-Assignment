```python
#Develop a python program to implement a Rule-Based Expert System for Student Performance Evaluation. The system should take inputs such as attendance, marks, assignment performance, and internal assessment and use IF THEN rules to classify the student as Excellent, Good, Average, or Poor. 

# Rule-Based Expert System for Student Performance Evaluation

print("   STUDENT PERFORMANCE EVALUATION SYSTEM")

# Taking input from the user
attendance = float(input("Enter Attendance Percentage: "))
marks = float(input("Enter Marks Percentage: "))
assignment = float(input("Enter Assignment Performance percentage: "))
internal = float(input("Enter Internal Assessment percentage: "))

# Rule-based classification

# Rule 1: Excellent
if (attendance >= 90 and marks >= 85 and
        assignment >= 85 and internal >= 85):

    performance = "Excellent"
    reason = "The student has excellent attendance, marks, assignments, and internal assessment."

# Rule 2: Good
elif (attendance >= 75 and marks >= 70 and
      assignment >= 70 and internal >= 70):

    performance = "Good"
    reason = "The student has good overall academic performance."

# Rule 3: Average
elif (attendance >= 60 and marks >= 50 and
      assignment >= 50 and internal >= 50):

    performance = "Average"
    reason = "The student has satisfactory performance but needs some improvement."

# Rule 4: Poor
else:

    performance = "Poor"
    reason = "The student needs significant improvement in academic performance."

# Displaying the result
print("EVALUATION RESULT")

print("Attendance          :", attendance, "%")
print("Marks               :", marks, "%")
print("Assignment          :", assignment, "%")
print("Internal Assessment :", internal, "%")

print("Performance Level   :", performance)
print("Reason              :", reason)
```

```python
def get_user_input(symptom_name):
    """Prompt the user for a yes/no symptom response."""
    while True:
        response = input(f"Do you have {symptom_name}? (yes/no): ").strip().lower()
        if response in ["yes", "y"]:
            return True
        elif response in ["no", "n"]:
            return False
        print("Invalid input. Please enter 'yes' or 'no'.")


def diagnose(fever, cough, headache, sore_throat):
    """Diagnoses possible illnesses based on symptom combinations."""
    # Rule 1: High symptom overlap -> Influenza (Flu)
    if fever and cough and headache and sore_throat:
        return (
            "Influenza (Flu)",
            "High",
            "Rest, stay hydrated, and consult a doctor if fever persists.",
        )

    # Rule 2: Upper respiratory focus without high systemic symptoms -> Common Cold
    elif cough and sore_throat and not headache:
        return (
            "Common Cold",
            "Moderate",
            "Over-the-counter cold relief, fluids, and throat lozenges.",
        )

    # Rule 3: Localized throat inflammation with fever -> Strep Throat
    elif fever and sore_throat and not cough:
        return (
            "Strep Throat / Tonsillitis",
            "Moderate",
            "Consider a throat swab or medical consultation for antibiotics.",
        )

    # Rule 4: Head pain and fever without respiratory tract symptoms -> Systemic infection / Migraine
    elif fever and headache and not cough and not sore_throat:
        return (
            "Viral Fever or Migraine with Fever",
            "Moderate",
            "Monitor body temperature, rest in a dark room, and hydrate.",
        )

    # Rule 5: Isolated persistent cough -> Bronchial or Allergic reaction
    elif cough and not fever and not headache and not sore_throat:
        return (
            "Allergic Bronchitis / Mild Cough",
            "Low",
            "Monitor for respiratory irritation or environmental allergens.",
        )

    # Rule 6: Isolated headache -> Tension headache
    elif headache and not fever and not cough and not sore_throat:
        return (
            "Tension Headache",
            "Low",
            "Rest, reduce screen exposure, and maintain adequate hydration.",
        )

    # Rule 7: No symptoms present
    elif not fever and not cough and not headache and not sore_throat:
        return (
            "Healthy / No Apparent Illness",
            "None",
            "No symptoms indicated.",
        )

    # Fallback: Inconclusive combinations
    else:
        return (
            "Undetermined Viral Syndrome",
            "Low to Moderate",
            "Symptoms do not match a single pattern. Consult a general physician.",
        )


def main():
    print("=== Rule-Based Medical Diagnosis Expert System ===")
    print("Answer the following questions to check possible conditions:\n")

    fever = get_user_input("a fever")
    cough = get_user_input("a cough")
    headache = get_user_input("a headache")
    sore_throat = get_user_input("a sore throat")

    condition, severity, advice = diagnose(fever, cough, headache, sore_throat)

    print("\n--- Diagnostic Summary ---")
    print(f"Reported Symptoms   : Fever={fever}, Cough={cough}, Headache={headache}, Sore Throat={sore_throat}")
    print(f"Possible Condition  : {condition}")
    print(f"Likelihood / Urgency: {severity}")
    print(f"Recommendation      : {advice}")


if __name__ == "__main__":
    main()
```
