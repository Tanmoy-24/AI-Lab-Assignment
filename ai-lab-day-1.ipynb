```python
import matplotlib.pyplot as plt

# Data
epochs = [1,2,3,4,5,6,7,8,9]
train_loss = [0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2, 0.1]
val_loss = [0.95, 0.75, 0.58, 0.48, 0.42, 0.38, 0.31, 0.25, 0.11]

# Create figure
plt.figure(figsize=(8, 5))

# Plot lines
plt.plot(epochs, train_loss, label='Training Loss', color='blue')
plt.plot(epochs, val_loss, label='Validation Loss', color='red')

# Title and labels
plt.title('AI Model Training Performance', fontsize=14, fontweight='bold')
plt.xlabel('Epochs (Training Iterations)', fontsize=12)
plt.ylabel('Loss (Error Rate)', fontsize=12)

# Grid and legend
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)

# Show plot
plt.show()
```
