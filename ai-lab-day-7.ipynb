```python
import math

def minimax_with_path(depth, node_index, is_max, scores, height):
    # Base case: leaf node reached
    if depth == height:
        return scores[node_index], [scores[node_index]]

    if is_max:
        left_val, left_path = minimax_with_path(depth + 1, node_index * 2, False, scores, height)
        right_val, right_path = minimax_with_path(depth + 1, node_index * 2 + 1, False, scores, height)
        
        if left_val > right_val:
            return left_val, [scores[node_index]] + left_path
        else:
            return right_val, [scores[node_index]] + right_path
    else:
        left_val, left_path = minimax_with_path(depth + 1, node_index * 2, True, scores, height)
        right_val, right_path = minimax_with_path(depth + 1, node_index * 2 + 1, True, scores, height)
        
        if left_val < right_val:
            return left_val, [scores[node_index]] + left_path
        else:
            return right_val, [scores[node_index]] + right_path

if __name__ == "__main__":
    leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]
    tree_height = int(math.log(len(leaf_values), 2))
    
    optimal_score, optimal_path = minimax_with_path(0, 0, True, leaf_values, tree_height)
    print(f"The optimal score for the MAX player is: {optimal_score}")
    print(f"The optimal path is: {' -> '.join(map(str, optimal_path))}")
```
