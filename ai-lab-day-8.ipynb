```python
# Write a python program to implement the minimax algorithm for a two-player game.

# 1. Implement the Minimax algorithm using recursion.
# 2. Consider MAX and MIN players.
# 3. Use the following leaf-node values:

#    3, 5, 2, 9, 12, 5, 23, 23

# 4. Determine the optimal score for the MAX player.
# 5. Display the final optimal score as output.
```

```python
import math

def minimax(depth, node_index, is_maximizing, values, target_depth):
    # Base case: reached leaf node level
    if depth == target_depth:
        return values[node_index]

    if is_maximizing:
        # MAX player aims to maximize the score
        return max(
            minimax(depth + 1, node_index * 2, False, values, target_depth),
            minimax(depth + 1, node_index * 2 + 1, False, values, target_depth)
        )
    else:
        # MIN player aims to minimize the score
        return min(
            minimax(depth + 1, node_index * 2, True, values, target_depth),
            minimax(depth + 1, node_index * 2 + 1, True, values, target_depth)
        )

# Leaf-node values
values = [3, 5, 2, 9, 12, 5, 23, 23]

# Calculate tree depth dynamically based on the number of leaf nodes (log2(8) = 3)
tree_depth = int(math.log2(len(values)))

# Run the Minimax algorithm starting at the root (depth 0, index 0, MAX player's turn)
optimal_score = minimax(0, 0, True, values, tree_depth)

# Output
print(f"Leaf Node Values: {values}")
print(f"Optimal score for MAX player: {optimal_score}")
```
